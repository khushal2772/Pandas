```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv(r'anime.csv')
```

```python
df.head()
```

```python
df.loc[4]['Title']
```

```python
def extract_ep(txt):
    check = False
    txt1 = ""
    for i in txt:
        if i == ')':
            break
        if check:
            txt1 += i
        if i == '(':
            check = True
    return txt1
```

```python
df['ep'] = df['Title'].apply(extract_ep)
df
```

```python
df['ep'] = df['ep'].str.replace(" eps","")
```

```python
df['ep'] = df['ep'].astype(int)
```

```python
def extract_time(txt):
    check = False
    data = ''
    k = 19
    for i in txt:
        if check and k > 0:
            data += i
            k -= 1
        if k <= 0:
            return data
            break
        if i == ')':
            check = True
        
```

```python
df['Time'] = df['Title'].apply(extract_time)
```

```python
df
```

```python
def extract_members(txt):
    check = False
    check1 = False
    k = 0
    data = ''
    
    for i in txt:
        if k > 19:
            check1 = True
            data += i
        if i == 's' and check1:
            return data
        if i == ')':
            check = True
        if check:
            k += 1
        
        
```

```python
df['Title'].apply(extract_members)
```
